```python
# solo con RateCode 4 y 5
tarifas_especiales = datos_crudos[datos_crudos['RatecodeID'].isin([4, 5])]

resumen_tarifas = tarifas_especiales.groupby('RatecodeID')['fare_amount'].agg(['min', 'max', 'mean', 'count'])

resumen_tarifas.columns = ['Mínimo ($)', 'Máximo ($)', 'Promedio ($)', 'Total Viajes']
```

```python
# Definimos la lista de columnas que componen el total
columnas_suma = [
    'fare_amount', 'extra', 'mta_tax', 'tip_amount', 'tolls_amount',
    'improvement_surcharge', 'congestion_surcharge', 'Airport_fee'
]

# Creamos una serie con la suma de esas columnas para cada fila
# .sum(axis=1) suma de forma horizontal (por fila)
suma_calculada = datos_crudos[columnas_suma].sum(axis=1)

discrepancias_total = datos_crudos[round(suma_calculada, 2) != round(datos_crudos['total_amount'], 2)]

# Ver cuántos errores de suma encontramos
print(f"Registros con descuadre en el total: {len(discrepancias_total)}")
```
